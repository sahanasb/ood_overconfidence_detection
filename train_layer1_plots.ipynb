# Layer 1 fingerprint plots
Training fingerprints: NSL-KDD vs UNSW-NB15

Exploratory plots for Layer-1 fingerprints.

This script uses the Layer-2 TRAINING fingerprint pools:
    - NSL-KDD: nsl_detector_train_fingerprints.csv  (ID)
    - UNSW-NB15: unsw_portion1_fingerprints.csv    (OOD)

```python
# load fingerprints
lr_nsl = pd.read_csv(
    "model_outputs/logistic_regression_fingerprints/nsl_detector_train_fingerprints.csv"
)
lr_unsw = pd.read_csv(
    "model_outputs/logistic_regression_fingerprints/unsw_portion1_fingerprints.csv"
)

svm_nsl = pd.read_csv(
    "model_outputs/svm_fingerprints/nsl_detector_train_fingerprints.csv"
)
svm_unsw = pd.read_csv(
    "model_outputs/svm_fingerprints/unsw_portion1_fingerprints.csv"
)

mlp_nsl = pd.read_csv(
    "model_outputs/mlp_fingerprints/nsl_detector_train_fingerprints.csv"
)
mlp_unsw = pd.read_csv(
    "model_outputs/mlp_fingerprints/unsw_portion1_fingerprints.csv"
)
```

```python
# Confidence

fig, axes = plt.subplots(1, 3, figsize=(18, 5), sharex=True, sharey=True)

# Logistic Regression

axes[0].hist(
    lr_nsl["confidence"],
    bins=30,
    density=True,
    alpha=0.5,
    label="NSL-KDD (ID)"
)

axes[0].hist(
    lr_unsw["confidence"],
    bins=30,
    density=True,
    alpha=0.5,
    label="UNSW-NB15 (OOD)"
)

axes[0].set_title("Logistic Regression")
axes[0].set_xlabel("Confidence")
axes[0].set_ylabel("Density")
axes[0].legend()

# SVM  
axes[1].hist(
    svm_nsl["confidence"],
    bins=30,
    density=True,
    alpha=0.5,
    label="NSL-KDD (ID)"
)

axes[1].hist(
    svm_unsw["confidence"],
    bins=30,
    density=True,
    alpha=0.5,
    label="UNSW-NB15 (OOD)"
)

axes[1].set_title("SVM")
axes[1].set_xlabel("Confidence")
axes[1].legend()

# MLP
  
axes[2].hist(
    mlp_nsl["confidence"],
    bins=30,
    density=True,
    alpha=0.5,
    label="NSL-KDD (ID)"
)

axes[2].hist(
    mlp_unsw["confidence"],
    bins=30,
    density=True,
    alpha=0.5,
    label="UNSW-NB15 (OOD)"
)

axes[2].set_title("MLP")
axes[2].set_xlabel("Confidence")
axes[2].legend()


# Overall title
fig.suptitle(
    "Confidence Distribution: In-Distribution vs Out-of-Distribution",
    fontsize=16
)

plt.tight_layout()
plt.show()
```

```python
# ECE

models = ["Logistic Regression", "SVM", "MLP"]

ece_nsl = [
    0.11655509726856148,
    0.1366565790979187,
    0.13717489130370067
]

ece_unsw = [
    0.36236511961248896,
    0.5985694968184073,
    0.43970030099908963
]

x = np.arange(len(models))
width = 0.35

plt.figure(figsize=(8, 6))

plt.bar(
    x - width/2,
    ece_nsl,
    width,
    label="NSL-KDD"
)

plt.bar(
    x + width/2,
    ece_unsw,
    width,
    label="UNSW-NB15"
)

plt.xticks(x, models)
plt.ylabel("Expected Calibration Error (ECE)")
plt.title("ECE Across Layer-1 Models")
plt.legend()

plt.show()
```
